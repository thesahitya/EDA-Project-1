#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
````

````python
heat = df.copy()

heat['Medal'] = heat['Medal'].astype('category').cat.codes
heat['Gender'] = heat['Gender'].astype('category').cat.codes
heat['Sport'] = heat['Sport'].astype('category').cat.codes
heat['Country'] = heat['Country'].astype('category').cat.codes

plt.figure(figsize=(6,5))
sns.heatmap(heat[['Sport','Country','Gender','Medal']].corr(),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
````

##### 1. Why did you pick the specific chart?

A correlation heatmap was chosen to quickly visualize the relationships between multiple variables. It helps identify whether any variables are positively, negatively, or weakly correlated.

##### 2. What is/are the insight(s) found from the chart?

1. The heatmap shows that most variables have weak correlations, indicating they are largely independent of each other.
2. No strong positive or negative relationships are observed, suggesting that factors like Sport, Country, Gender, and Medal do not have a simple linear relationship in this dataset.
3. This helps understand that Olympic medal outcomes are influenced by multiple factors rather than a single variable.

#### Chart - 15 - Pair Plot
